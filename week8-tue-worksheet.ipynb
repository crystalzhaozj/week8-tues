### Q8: Time frame for a given country

Write a function that given a country name prints out the start and end date of their rankings.

Once you have created the function, call it for a few countries to test it.

```python
type(df['week'])
```

```python
df['week'] = pd.to_datetime(df['week'])
```

```python
df['week'].min
```

This returns a whole series...

```python
def show_timeframe(data, country): 
    country_df = df[df['country_name'] == country]
    start = country_df['week'].min()
    end = country_df['week'].max()
    print(f"Netflix rankings for {country} exist between {start} and {end}.")
```

```python
show_timeframe(df,'Chile')
```

***New function: rank2score***

```python
def rank2score(rank):
    if rank == 1:
        score = 25
    elif rank == 2:
        score = 18
    elif rank == 3: 
        score = 15
    elif rank == 4:
        score = 12
    elif rank == 5: 
        score = 10
    elif rank == 6:
        score = 8 
    elif rank == 7: 
        score = 6
    elif rank == 8:
        score = 4
    elif rank == 9:
        score = 2
    elif rank == 10:
        score = 1
    return score
```

```python
df.head()
```

```python
df['score'] = df['weekly_rank'].apply(rank2score)
df.head()
```

Removing the useless columns:

```python
df = df.drop(columns=['country_iso2', 'category', 'season_title'])
```

```python
df.head()
```

```python
lr_df = df[df['show_title']== 'Lovely Runner']
lr_df.shape
```

```python
lr_df.head()
```

There are 51 rows with Lovely Runner.

Total score:

```python
lr_df_stats = lr_df.groupby('country_name')['score'].sum().rename('total_score').reset_index()
lr_df_stats
```

Total weeks:

```python
lr_df_stats['total_weeks']= lr_df.groupby('country_name')['week'].count().rename('total_weeks').reset_index()['total_weeks']
lr_df_stats
```

Rankings_list

```python
lr_df_stats['rankings_list'] = lr_df.groupby('country_name')['weekly_rank'].agg(list).rename('rankings_list').reset_index()['rankings_list']
lr_df_stats
```

```python
lr_df_stats.sort_values(by='total_score', ascending=False)
```
